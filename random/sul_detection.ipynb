```python
import pandas as pd
import os
```

```python
folder = [f for f in os.listdir("../resources/matching") if f.endswith('.csv')]
```

```python
result = []

for csv in folder:
  df = pd.read_csv(f"../resources/matching/{csv}")
  sum_detected = 0
  sum_not_detected = 0
  for index, row in df.iterrows():
    if row['title'] == row['predicted_title']:
      sum_detected += 1
    else:
      sum_not_detected += 1
      
  result.append({
    'name': csv,
    'detected': sum_detected,
    'not_detected': sum_not_detected
  })
pd.DataFrame(result)
    
```
